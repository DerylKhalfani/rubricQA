```python
import json
import pandas as pd

with open("../../data/abcd_pretty.json") as f:
    data = json.load(f)

print(type(data))
if isinstance(data, dict):
    print(data.keys())
```

```python
rows = []

convos = data['train']

for c in convos:
    rows.append({
        "convo_id": c["convo_id"],
        "flow": c["scenario"]["flow"],
        "subflow": c["scenario"]["subflow"],
        "n_turns": len(c["original"]),
    })

df = pd.DataFrame(rows)
df.head()
```

```python
def show(convos: list[dict], convo_id: int):
    """To show conversation between agent and user"""

    for convo in convos:
        if convo['convo_id'] == convo_id:
            print(f"\nConversation: {convo_id} \n")
            
            for speaker, text in convo["original"]:
                print(f"{speaker}: {text}")
            return

    print(f"No conversation with id {convo_id}")

# show(convos, 3592)

```

```python
for flow, group in df.groupby("flow"):
    print(f"=== {flow} ===")
    for convo_id in group["convo_id"].head(2):
        show(convos, int(convo_id))
        print("-" * 40)
```
